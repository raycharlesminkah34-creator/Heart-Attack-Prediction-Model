```python
#Convert type str to object int cat variables
str_cols = df.select_dtypes(include="string").columns
df[str_cols] = df[str_cols].astype(object)
```

```python
df.info()
```

```python
#Data type grouping 
binary_cols = [col for col in df.columns if df[col].nunique() == 2]
#cat_cols = [col for col in df.columns if (col not in binary_cols) & (col not in continuous_cols) & (col != "Outcome")]
#continuous_cols = [col for col in df.columns if df[col].nunique() > 25]

binary_cols
```

```python
binary_maps = {
    'Gender': {"Male": 1, "Female": 0},
    'Medication': {"Yes": 1, "No": 0},
    'ExerciseInducedAngina': {"Yes": 1, "No": 0},
    'Outcome': {"No Heart Attack": 0, "Heart Attack": 1}
    
}
```

```python
#Binary Encoding
for col, mapping in binary_maps.items():
    df[col] = df[col].map(mapping)
```

```python
df.head(10)
```

```python
df.dtypes
```

```python
thal_map = {
    "Normal": 0,
    "Reversible defect": 1,
    "Fixed defect": 2
}

slope_map = {
    "Upsloping": 0,
    "Flat": 1, 
    "Downsloping": 2
}

ecg_map = {
    "Normal": 0,
    "ST-T abnormality": 1,
    "LV hypertrophy": 2
}
```

```python
#encoding roadmap{identifying the cardinality of variables to determine the encoding method}
for col in df.select_dtypes(include="object").columns:
    print(f"{col}: {df[col].nunique()}")
```

```python
#Ordinal encoding 
df['Thalassemia'] = df['Thalassemia'].map(thal_map)    
df['Slope'] = df['Slope'].map(slope_map)    
df['ECGResults'] = df['ECGResults'].map(ecg_map)    
```

```python
df['ECGResults'].unique()
```

```python
#One-hot encoding
nominal_cols = ['Diet', 'Ethnicity', 'EducationLevel', 'Residence', 'EmploymentStatus', 'MaritalStatus', 'ChestPainType']
df_encoded = pd.get_dummies(df, columns=nominal_cols, drop_first=True, dtype=int)
```
